```python
import os
import shutil

# 1. Tự động tìm vị trí chính xác của file run_pipeline.py trong /kaggle/input
target_file = "run_pipeline.py"
found_src_dir = None

for root, dirs, files in os.walk("/kaggle/input"):
    if target_file in files:
        found_src_dir = root
        break

dest_dir = "/kaggle/working/GRACE"

if found_src_dir:
    print(f"[*] Đã tìm thấy thư mục chứa mã nguồn tại: {found_src_dir}")
    if os.path.exists(dest_dir):
        shutil.rmtree(dest_dir)
    # Copy chính xác thư mục chứa file run_pipeline.py
    shutil.copytree(found_src_dir, dest_dir)
    print(f"✓ [OK] Đã sao chép toàn bộ mã nguồn sang: {dest_dir}")
else:
    print("[!] ERROR: Không tìm thấy file run_pipeline.py trong /kaggle/input. Vui lòng kiểm tra lại dataset đính kèm!")

# 2. Chuyển thư mục làm việc về /kaggle/working/GRACE
os.chdir(dest_dir)
print(f"✓ [OK] Thư mục làm việc hiện tại: {os.getcwd()}")
```
